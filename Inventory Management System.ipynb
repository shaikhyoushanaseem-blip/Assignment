# *Store Information (Task 1)*

```python
store_name = input("Enter store name: ")
manager_name = input("Enter manager name: ")

store_code = "ELEC-2026-KHI"

category_code = store_code[0:4]
year = store_code[5:9]
city_code = store_code[10:13]

print("STORE INFORMATION")
print("Store Name    :", store_name)
print("Manager Name  :", manager_name)
print("Store Code    :", store_code)
print("Category Code :", category_code)
print("Year          :", year)
print("City Code     :", city_code)
```

# *Create the Inventory (Task 2)*

```python
products = ["Laptop", "Mouse", "Keyboard", "Monitor", "Headphones"]

print("INITIAL INVENTORY")
print("Products      :", products)
print("Total Products:", len(products))
```

# *Add New Product (Task 3)*

```python
new_product = input("Enter a new product: ")
products.append(new_product)
products.extend(["Webcam", "USB Hub", "Speaker"])
print("After Add Products")
print("-" * 55)
print("Inventory:", products)
```

# *Insert a Product (Task 4)*

```python
monitor_position = products.index("Monitor")
products.insert(monitor_position + 1, "Printer")

print("\nAFTER INSERTING PRINTER")
print("Monitor Position :", monitor_position)
print("Inventory        :", products)
```

# *Remove a Discontinued Product (Task 5)*

```python
removed_product = products.pop("Headphones")
print("\nAFTER REMOVING DISCONTINUED PRODUCT")
print("-" * 55)
print("Removed Product: Headphones")
print("Inventory      :", products)
```

# *Process a Recent Shipment (Task 6)*

```python
shipment = ["Tablet", "Smart Watch", "Power Bank"]
products.extend(shipment)
removed_product = products.pop()
print("RECENT SHIPMENT")
print("Shipment Added :", shipment)
print("Removed Last Product:", removed_product)
print("Inventory      :", products)
```

# *Product Information (Task 7)*

```python
laptop_position = products.index("Laptop")
mouse_count = products.count("Mouse")

print("PRODUCT INFORMATION")
print("Laptop Position:", laptop_position)
print("Mouse Count    :", mouse_count)
```

# *Sort The Inventory (Task 8)*

```python
products.sort()
sorted_az = products.copy()

print("SORTED INVENTORY (A-Z)")
print(sorted_az)

products.sort(reverse=True)
sorted_za = products.copy()

print("SORTED INVENTORY (Z-A)")
print(sorted_za)
```

# *Reverse The Inventory (Task 9)*

```python
products.reverse()
reversed_inventory = products.copy()

print("\nREVERSED INVENTORY")

print(reversed_inventory)
```

# *Create an Inventory Backup (Task 10)*

```python
backup_products = products.copy()

print("\nINVENTORY BACKUP")
print("Current Inventory:", products)
print("Backup Inventory :", backup_products)
```

# *Clear Temporary Inventory (Task 11)*

```python
temporary_inventory = ["Test Laptop", "Test Mouse", "Test Keyboard"]
temporary_inventory.clear()

print("\nTEMPORARY INVENTORY AFTER CLEAR")
print(temporary_inventory)
```
